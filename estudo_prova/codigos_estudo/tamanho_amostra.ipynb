```python
import statsmodels.stats.api as sms

# Parâmetros de entrada
p_base = 0.05            # 5% de taxa de conversão atual
mde_relativo = 0.10      # 10% de ganho relativo desejado
alpha = 0.05              # 95% de confiança
power = 0.80              # 80% de poder

# Cálculo do P2 (taxa esperada no grupo de tratamento)
p_tratamento = p_base * (1 + mde_relativo)

# Tamanho do efeito (Cohen's h para proporções)
effect_size = sms.proportion_effectsize(p_base, p_tratamento)

# Cálculo do n por grupo
n_por_grupo = sms.NormalIndPower().solve_power(
    effect_size=effect_size, 
    alpha=alpha, 
    power=power, 
    ratio=1.0
)

print(f"Tamanho de amostra por variante: {int(round(n_por_grupo))}")
print(f"Amostra total do experimento: {int(round(n_por_grupo * 2))}")
```
